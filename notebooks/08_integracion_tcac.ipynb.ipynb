```python
# ------------------------------------------------------------
# 4. Definir columnas TCAC
# ------------------------------------------------------------
# Ajustar si los nombres reales cambian en tu archivo

col_tcac_codigo = "codigo"
col_tcac_nombre = "nombre_alimento"

print("Columna código TCAC:", col_tcac_codigo)
print("Columna nombre TCAC:", col_tcac_nombre)
```

```python
# ------------------------------------------------------------
# 5. Funciones de limpieza textual
# ------------------------------------------------------------

import unicodedata
import re
import pandas as pd

def quitar_tildes(texto):
    if pd.isna(texto):
        return ""

    texto = str(texto)

    texto = unicodedata.normalize("NFD", texto)

    texto = "".join(
        c for c in texto
        if unicodedata.category(c) != "Mn"
    )

    return texto


def limpiar_texto_matching(texto):
    if pd.isna(texto):
        return ""

    texto = str(texto).lower()

    texto = quitar_tildes(texto)

    texto = re.sub(r"[^a-z0-9\s]", " ", texto)

    texto = re.sub(r"\s+", " ", texto).strip()

    return texto
```

```python
# ------------------------------------------------------------
# 6. Limpiar ingredientes del pipeline
# ------------------------------------------------------------

import pandas as pd
import unicodedata
import re

ingredientes["ingredient_standardized_clean"] = (
    ingredientes["ingredient_standardized"]
    .apply(limpiar_texto_matching)
)

ingredientes.head()
```

```python
# ------------------------------------------------------------
# 7. Limpiar nombres de alimentos TCAC
# ------------------------------------------------------------

import pandas as pd
import unicodedata
import re

tcac["tcac_food_clean"] = (
    tcac[col_tcac_nombre]
    .apply(limpiar_texto_matching)
)

tcac.head()
```

```python
# ------------------------------------------------------------
# 8. Revisar ingredientes únicos antes del matching
# ------------------------------------------------------------

print("Ingredientes estandarizados únicos:")
print(
    ingredientes["ingredient_standardized_clean"]
    .nunique()
)

display(
    ingredientes["ingredient_standardized_clean"]
    .value_counts()
    .head(50)
)
```

```python
# ------------------------------------------------------------
# 9. Preparar lista de alimentos TCAC
# ------------------------------------------------------------

tcac_foods = tcac["tcac_food_clean"].dropna().unique().tolist()

print("Número de alimentos TCAC:", len(tcac_foods))
```

```python
# ------------------------------------------------------------
# 10. Función de matching híbrido
# ------------------------------------------------------------

import pandas as pd
from fuzzywuzzy import fuzz, process

def buscar_match_tcac(ingrediente, threshold=80):

    if pd.isna(ingrediente) or ingrediente == "":
        return None, None, None, "sin_match"

    ingrediente = str(ingrediente).strip()

    # --------------------------------------------------------
    # 1. Matching exacto
    # --------------------------------------------------------

    exacto = tcac[
        tcac["tcac_food_clean"] == ingrediente
    ]

    if not exacto.empty:

        fila = exacto.iloc[0]

        return (
            fila[col_tcac_codigo],
            fila[col_tcac_nombre],
            100,
            "exact"
        )

    # --------------------------------------------------------
    # 2. Fuzzy matching token_sort_ratio
    # --------------------------------------------------------

    resultado = process.extractOne(
        ingrediente,
        tcac_foods,
        scorer=fuzz.token_sort_ratio
    )

    if resultado:

        mejor_match, score, _ = resultado

        if score >= threshold:

            fila = tcac[
                tcac["tcac_food_clean"] == mejor_match
            ].iloc[0]

            return (
                fila[col_tcac_codigo],
                fila[col_tcac_nombre],
                score,
                "fuzzy_token_sort"
            )

    # --------------------------------------------------------
    # 3. Fuzzy matching partial_ratio
    # --------------------------------------------------------

    resultado_partial = process.extractOne(
        ingrediente,
        tcac_foods,
        scorer=fuzz.partial_ratio
    )

    if resultado_partial:

        mejor_match, score, _ = resultado_partial

        if score >= threshold:

            fila = tcac[
                tcac["tcac_food_clean"] == mejor_match
            ].iloc[0]

            return (
                fila[col_tcac_codigo],
                fila[col_tcac_nombre],
                score,
                "fuzzy_partial"
            )

    # --------------------------------------------------------
    # 4. Sin match
    # --------------------------------------------------------

    return None, None, None, "sin_match"
```

```python
# Re-defining the matching function after installing fuzzywuzzy

import pandas as pd
from fuzzywuzzy import fuzz, process

def buscar_match_tcac(ingrediente, threshold=80):

    if pd.isna(ingrediente) or ingrediente == "":
        return None, None, None, "sin_match"

    ingrediente = str(ingrediente).strip()

    exacto = tcac[
        tcac["tcac_food_clean"] == ingrediente
    ]

    if not exacto.empty:

        fila = exacto.iloc[0]

        return (
            fila[col_tcac_codigo],
            fila[col_tcac_nombre],
            100,
            "exact"
        )

    resultado_token = process.extractOne(
        ingrediente,
        tcac_foods,
        scorer=fuzz.token_sort_ratio
    )

    if resultado_token and len(resultado_token) == 3: # Added length check

        mejor_match, score, _ = resultado_token

        if score >= threshold:

            fila = tcac[
                tcac["tcac_food_clean"] == mejor_match
            ].iloc[0]

            return (
                fila[col_tcac_codigo],
                fila[col_tcac_nombre],
                score,
                "fuzzy_token_sort"
            )

    resultado_partial = process.extractOne(
        ingrediente,
        tcac_foods,
        scorer=fuzz.partial_ratio
    )

    if resultado_partial and len(resultado_partial) == 3: # Added length check

        mejor_match, score, _ = resultado_partial

        if score >= threshold:

            fila = tcac[
                tcac["tcac_food_clean"] == mejor_match
            ].iloc[0]

            return (
                fila[col_tcac_codigo],
                fila[col_tcac_nombre],
                score,
                "fuzzy_partial"
            )

    return None, None, None, "sin_match"
```

```python
# Re-applying the matching function after its definition

import pandas as pd
from tqdm.auto import tqdm

tqdm.pandas()

resultado_matching = ingredientes[
    "ingredient_standardized_clean"
].progress_apply(
    buscar_match_tcac
)

(
    ingredientes["tcac_codigo"],
    ingredientes["tcac_nombre"],
    ingredientes["match_score"],
    ingredientes["match_method"]
) = zip(*resultado_matching)

ingredientes.head()
```

```python
# ------------------------------------------------------------
# 15. Actualizar JSON con información TCAC
# ------------------------------------------------------------

import json

input_json = "data/processed/recetas_estructura_json_convertida.json"

with open(input_json, "r", encoding="utf-8") as f:
    recetas_json = json.load(f)

print("JSON cargado:", len(recetas_json))
```

```python
# ------------------------------------------------------------
# 16. Crear mapa de matching TCAC
# ------------------------------------------------------------

mapa_tcac = {}

for _, row in ingredientes.iterrows():

    clave = (
        str(row["recipe_id"]),
        int(row["ingredient_id"])
    )

    mapa_tcac[clave] = {
        "tcac_code": None if pd.isna(row["tcac_codigo"]) else row["tcac_codigo"],
        "tcac_food_name": None if pd.isna(row["tcac_nombre"]) else row["tcac_nombre"],
        "match_score": None if pd.isna(row["match_score"]) else float(row["match_score"]),
        "match_method": row["match_method"]
    }
```

```python
# ------------------------------------------------------------
# 17. Actualizar estructura JSON
# ------------------------------------------------------------

for receta in recetas_json:

    recipe_id = str(receta["recipe_id"])

    for ingrediente in receta["ingredients"]:

        ingredient_id = ingrediente["ingredient_id"]

        clave = (recipe_id, ingredient_id)

        if clave in mapa_tcac:

            ingrediente["tcac_match"] = mapa_tcac[clave]

    receta["pipeline_stage"] = "08_integracion_tcac"
    receta["structure_version"] = "1.3"

print("JSON actualizado correctamente.")
```

# Integración con la TCAC

En esta etapa se realiza la integración entre los ingredientes estandarizados del pipeline y la Tabla de Composición de Alimentos Colombianos (TCAC).

El objetivo es asociar cada ingrediente procesado con el alimento más similar disponible en la TCAC para posteriormente calcular el perfil nutricional de cada receta.

La integración se realiza mediante una estrategia híbrida:

- coincidencia exacta,
- coincidencia normalizada,
- fuzzy matching utilizando RapidFuzz.

Esta etapa utiliza como entrada:

- `data/interim/recetas_unidades_convertidas.csv`
- `data/raw/tcac/tcac.csv`

`tcac_match`
