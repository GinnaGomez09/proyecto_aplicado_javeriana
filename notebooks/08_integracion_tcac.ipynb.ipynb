```python
# ------------------------------------------------------------
# 5. Funciones de limpieza textual
# ------------------------------------------------------------

import unicodedata
import re
import pandas as pd

def quitar_tildes(texto):
    if pd.isna(texto):
        return ""

    texto = str(texto)

    texto = unicodedata.normalize("NFD", texto)

    texto = "".join(
        c for c in texto
        if unicodedata.category(c) != "Mn"
    )

    return texto


def limpiar_texto_matching(texto):
    if pd.isna(texto):
        return ""

    texto = str(texto).lower()

    texto = quitar_tildes(texto)

    texto = re.sub(r"[^a-z0-9\s]", " ", texto)

    texto = re.sub(r"\s+", " ", texto).strip()

    return texto
```

```python
# ------------------------------------------------------------
# 6. Limpiar ingredientes del pipeline
# ------------------------------------------------------------

import pandas as pd
import unicodedata
import re

ingredientes["ingredient_standardized_clean"] = (
    ingredientes["ingredient_standardized"]
    .apply(limpiar_texto_matching)
)

ingredientes.head()
```

```python
# ------------------------------------------------------------
# 10. Función de matching híbrido
# ------------------------------------------------------------

import pandas as pd
from fuzzywuzzy import fuzz, process

def buscar_match_tcac(ingrediente, threshold=80):

    if pd.isna(ingrediente) or ingrediente == "":
        return None, None, None, "sin_match"

    ingrediente = str(ingrediente).strip()

    # --------------------------------------------------------
    # 1. Matching exacto
    # --------------------------------------------------------

    exacto = tcac[
        tcac["tcac_food_clean"] == ingrediente
    ]

    if not exacto.empty:

        fila = exacto.iloc[0]

        return (
            fila[col_tcac_codigo],
            fila[col_tcac_nombre],
            100,
            "exact"
        )

    # --------------------------------------------------------
    # 2. Fuzzy matching token_sort_ratio
    # --------------------------------------------------------

    resultado = process.extractOne(
        ingrediente,
        tcac_foods,
        scorer=fuzz.token_sort_ratio
    )

    if resultado:

        mejor_match, score, _ = resultado

        if score >= threshold:

            fila = tcac[
                tcac["tcac_food_clean"] == mejor_match
            ].iloc[0]

            return (
                fila[col_tcac_codigo],
                fila[col_tcac_nombre],
                score,
                "fuzzy_token_sort"
            )

    # --------------------------------------------------------
    # 3. Fuzzy matching partial_ratio
    # --------------------------------------------------------

    resultado_partial = process.extractOne(
        ingrediente,
        tcac_foods,
        scorer=fuzz.partial_ratio
    )

    if resultado_partial:

        mejor_match, score, _ = resultado_partial

        if score >= threshold:

            fila = tcac[
                tcac["tcac_food_clean"] == mejor_match
            ].iloc[0]

            return (
                fila[col_tcac_codigo],
                fila[col_tcac_nombre],
                score,
                "fuzzy_partial"
            )

    # --------------------------------------------------------
    # 4. Sin match
    # --------------------------------------------------------

    return None, None, None, "sin_match"
```
